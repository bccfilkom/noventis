> # Import Modul

```python
import sys
import os
import pandas as pd

module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)

from noventis.eda_auto import NoventisAutoEDA
print('Successfully imported modules!')
```

> # Prepare for Data Frame

```python
df_test = pd.read_csv('../dataset_for_examples/AmesHousing.csv')
```

> # Standard Comprehensive EDA

```python
analyzer_default = NoventisAutoEDA(data=df_test, target='SalePrice')
analyzer_default.run()
```

> # Focused Business Intelligence Dashboard

```python
analyzer_business = NoventisAutoEDA(data=df_test, target='SalePrice', personality='business')
analyzer_business.run(show_base_viz=False)
```

> # Rigorous Academic Statistical Report

```python
analyzer_academic = NoventisAutoEDA(df_test, target='SalePrice', personality='academic')
analyzer_academic.run()
```

> # All-in-One Report

```python
analyzer_full = NoventisAutoEDA(data=df_test, target='SalePrice', personality='all')
analyzer_full.run()
```
